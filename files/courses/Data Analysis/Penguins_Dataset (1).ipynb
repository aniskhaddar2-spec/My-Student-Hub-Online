```python
import matplotlib.pyplot as plt

plt.scatter(df["flipper_length_mm"], df["body_mass_g"])

plt.xlabel("Flipper length (mm)")
plt.ylabel("Body mass (g)")
plt.title("Flipper Length vs Body Mass")

plt.show()
```

```python
plt.scatter(
    df[df["species"] == "Adelie"]["flipper_length_mm"],
    df[df["species"] == "Adelie"]["body_mass_g"],
    label="Adelie"
)

plt.scatter(
    df[df["species"] == "Chinstrap"]["flipper_length_mm"],
    df[df["species"] == "Chinstrap"]["body_mass_g"],
    label="Chinstrap"
)

plt.scatter(
    df[df["species"] == "Gentoo"]["flipper_length_mm"],
    df[df["species"] == "Gentoo"]["body_mass_g"],
    label="Gentoo"
)

plt.xlabel("Flipper length (mm)")
plt.ylabel("Body mass (g)")
plt.title("Flipper Length vs Body Mass by Species")
plt.legend()

plt.show()
```
